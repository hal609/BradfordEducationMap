```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

base_domain = "https://reports.ofsted.gov.uk"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
}

# We will store each school as a dictionary inside this list
records = []

for urn in urns:
    print(f"\nSearching for URN: {urn}...")
    search_url = f"{base_domain}/search?q={urn}&location=&lat=&lon=&radius=&level_1_types=0"
    
    # 1. Fetch Search Page
    search_response = requests.get(search_url, headers=headers)
    search_soup = BeautifulSoup(search_response.content, 'html.parser')
    
    # 2. Extract First Search Result Link
    h3_tag = search_soup.find("h3", class_="search-result__title")
    first_result = h3_tag.find("a") if h3_tag else None
    
    if not first_result:
        print(f"No results found for URN {urn}.")
        # Append a row with missing data so you know this URN was checked
        records.append({
            "URN": urn,
            "OverallRating": None,
            "Quality of education": None,
            "Behaviour and attitudes": None,
            "Personal development": None,
            "Leadership and management": None
        })
        continue
        
    provider_path = first_result.get("href")
    provider_url = f"{base_domain}{provider_path}"
    print(f"Found provider page: {provider_url}")
    
    # 3. Fetch Provider Page
    provider_response = requests.get(provider_url, headers=headers)
    provider_soup = BeautifulSoup(provider_response.content, 'html.parser')

    # Extract Overall Rating
    overall_rating = None
    result_li_tag = provider_soup.find("li", class_="rating rating--selected heading--sub")
    if result_li_tag and result_li_tag.find("span"):
        overall_rating = result_li_tag.find("span").text.strip()
    
    # Start building the row dictionary for this school
    school_data = {
        "URN": urn,
        "OverallRating": overall_rating
    }

    if overall_rating is None:
        overall_other = provider_soup.find("div", class_="subjudgements__overall")
        if overall_other is not None:
            overall_rating = overall_other.find("strong").text

    # Extract Subjudgements
    rating_blocks = provider_soup.find_all("div", class_="subjudgements__rates__item")
    for block in rating_blocks:
        category_tag = block.find("p")
        rating_tag = block.find("strong")
        
        if category_tag and rating_tag:
            category = category_tag.text.strip().rstrip(':')
            rating = rating_tag.text.strip()
            school_data[category] = rating

    # Append this school's complete record to our list
    records.append(school_data)

# Convert the list of dictionaries directly into a DataFrame
df = pd.DataFrame(records)

# Display the final DataFrame
print("\n--- Final Ofsted Data ---")
print(df)
```
